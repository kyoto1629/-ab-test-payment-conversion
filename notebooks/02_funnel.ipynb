Единица метрики — попытка платежа; считаем переходы и абсолютный/относительный эффект.

Каждый ноутбук запускается независимо: **Restart Kernel → Run All**. Расчёты импортируются из `src/ab_test`. Исходные данные не изменяются.

```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src' / 'ab_test').is_dir())
sys.path.insert(0, str(ROOT / 'src'))
import pandas as pd
from IPython.display import display
pd.set_option('display.max_columns', None)
from ab_test.data import load_data, prepare_data
users, payments = load_data()
users_clean, payments_clean, audit = prepare_data(users, payments)
```

```python
from ab_test.metrics import calculate_funnel
funnel, step_rates = calculate_funnel(payments_clean)
display(funnel)
display(step_rates.round(2))
cr = funnel['step4_success'] / funnel['step1_opened']
display(cr.rename('conversion'))
print(f"Абсолютный прирост: {(cr['B']-cr['A'])*100:.2f} п.п.")
print(f"Относительный прирост: {(cr['B']/cr['A']-1)*100:.2f}%")
```
